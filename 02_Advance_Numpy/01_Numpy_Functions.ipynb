np.sort / return a sotred copy of an array

```python
a = np.random.randint(1,100,15)
b = np.random.randint(1,100,24).reshape(6,4)
print(a)
```

```python
print(b)
```

```python
print(np.sort(a))
print(np.sort(b)) # it will sort row wise automatically
# if you want your sort on colom wise
print(np.sort(b,axis=0))
```

```python
# if you your sorting on decending order
print(np.sort(a)[::-1])
```

np.concetenate() / Concetenate a sequence of arrays along an existing axis.

```python
c = np.arange(6).reshape(2,3)
d = np.arange(6,12).reshape(2,3)

print(c)
print(d)
```

```python
np.concatenate((c,d),axis=1)
```

np.where() / returns the indices of elements in an input array where the given condition is satisfied.

```python
a
```

```python
# find all the indices with value greater than 50.
np.where(a>50)
```

```python
# Replace all values which greater than 50 with 0. 
# np.where(condition, True,Flase) Syntax. 
np.where(a>50,0,a)
```
